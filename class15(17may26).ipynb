```python
def make_user_data(name,father_name,age,subject,last_classfication,experience,**other_info):
    profile = {}
    profile["name"]=name
    profile["father_name"]=father_name
    profile["age"]=age
    profile["subject"]=subject
    profile["last_classfication"]=last_classfication
    profile["experience"]=experience
    profile.update(other_info)
    return profile
```

```python
with open("guets_name.txt","w") as a:
    no=0
    while True:
        guets = input("Guest names: ")
        if guets!="q":
            a.write(f"{no}-{guets}\n")
            no+=1
        else:
            break
```
