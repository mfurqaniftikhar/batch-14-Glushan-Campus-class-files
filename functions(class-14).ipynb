```python
def disburse_saylary(saylary,units,years):
    commission = commission_calculator(units)
    bonous = bonous_calculaor(years)
    detuction = loan_detuctuion(saylary)
    total = saylary+commission+bonous
    tax = tax_calculator(total)
    gross_saylary = total-detuction-tax
    
    print(f"your total saylary is {gross_saylary}, and the tax is {tax} ")
```

```python
def commission_calculator(units):
    comm_amt = units*1000
    return comm_amt
```

```python
def bonous_calculaor(years):
    if years >=10:
        bonous =50000
    elif years >=8:
        bonous =40000
    elif years >=6:
        bonous =30000
    elif years >=4:
        bonous =20000
    elif years >=2:
        bonous =10000
    else:
        bonous = 5000
    return bonous
```

```python
def tax_calculator(total):
    if total<=500000:
        tax = total*0.0
    elif total<= 800000:
        tax = total*0.05
    elif total<= 1000000:
        tax = total*0.08
    elif total<= 1500000:
        tax = total*0.10
    elif total<= 1800000:
        tax = total*0.15
    else:
        tax = total*0.25
    return tax
```

```python
def loan_detuctuion(saylary):
    if saylary>=100000:
        deuc = saylary*0.30
    elif saylary>=70000:
        deuc = saylary*0.25
    elif saylary>=60000:
        deuc = saylary*0.20
    elif saylary>=40000:
        deuc = saylary*0.15
    elif saylary>=30000:
        deuc = saylary*0.10
    else:
        deuc= saylary*0.05
    return deuc
```
