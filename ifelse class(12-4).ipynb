# Ek program likho jisme ek variable time ho.
# Us variable ki value ke hisaab se program different messages print kare:

## Agar time "morning" ho → "Utho beta, class ka time ho gaya"
## Agar time "afternoon" ho → "Abhi tak kaam start nahi kiya?"
## Agar time "evening" ho → "Chai aur snacks ka time hogia hai "
## Agar time "night" ho → "Aaj se pakka parhai start karunga"
## Agar time "late night" ho → "Bas 5 min aur reel scroll... phir sota hoon"
## Agar time "midnight" ho → "Life decisions 3 baje raat ko hi aate hain"
## Agar in me se koi bhi value na ho → "Time hi samajh nahi aa raha"

```python
time = input("enter :morning, afternoon, evening, night, late night, midnight: ").lower()


if time == "morning":
    print(f"Utho beta, class ka time ho gaya, {time} ho rhi hai ")
elif time == "afternoon":
    print(f"Abhi tak kaam start nahi kiya?,  {time} ho rhi hai")
elif time == "evening":
    print(f"Chai aur snacks ka time hogia hai,  {time} ho rhi hai")
elif time == "night":
    print(f"Aaj se pakka parhai start karunga")
elif time == "late night":
    print("Bas 5 min aur reel scroll... phir sota hoon")
elif time == "midnight":
    print("Life decisions 3 baje raat ko hi aate hain")
else:
    print("Time hi samajh nahi aa raha")
```
