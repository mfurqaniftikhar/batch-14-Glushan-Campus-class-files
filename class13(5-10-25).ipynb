```python
# 1 - vowels
# 2 - punc
# 3 - upper 
# 4 - lower
# 5 - digit 
# 6 - space count 
# 7 - sentences
# 8 - text len 
# 9 - length
punc = "!#$%&'()*+,-./:;<=>?@[\]^_`{|}~"
vowel = "aeiouAEIOU"
def text_analyzer(text):
    length = len(text)
    sentences = len(text.split("."))
    words = (text.count(" ")+1)
    upper_case = 0
    lowercase = 0
    digit = 0
    puncutations = 0
    vowels = 0
    spaces_count = 0
    for char in text:
        if char.isupper():
            upper_case +=1
            if char in vowel:
                vowels+=1
                
        elif char.islower():
            lowercase+=1
            if char in vowel:
                vowels+=1
        elif char.isdigit():
            digit +=1
        elif char in punc:
            puncutations +=1
        elif char.isspace():
            spaces_count+=1
            
            
            
            
    print(f"""
              length of text : {length}
              sentences in text : {sentences}
              total words : {words}
              upper case words :{upper_case}
              lower case words :{lowercase}
              digits in text : {digit}'
              puncutations in text :{puncutations}
              vowels in text : {vowels}
              space counts :{spaces_count}
            """)  

```
