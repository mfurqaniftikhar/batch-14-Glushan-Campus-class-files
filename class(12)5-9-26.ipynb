```python
# +,-,*,/,**,//,%

while True:
    answer = 0
    num1 = (input("enter your number: "))
    if num1=="q":
        print("THANKS FOR USING SMIT BATCH-14 CALCULAOR")
        break
    num2 = (input("enter another number: "))
    if num2=="q":
        print("THANKS FOR USING SMIT BATCH-14 CALCULAOR")
        break
    op = (input("enter an operator from these +,-,*,/,**,//,%: "))
    if op=="q":
        print("THANKS FOR USING SMIT BATCH-14 CALCULAOR")
        break
    elif op =="+":
        answer = int(num1)+int(num2)
    elif op =="-":
        answer = int(num1)-int(num2)
    elif op =="*":
        answer = int(num1)*int(num2)
    elif op =="**":
        answer = int(num1)**int(num2)
    elif op =="/":
        answer = int(num1)/int(num2)
    elif op =="//":
        answer = int(num1)//int(num2)
    else:
        answer = int(num1)%int(num2)
        
    print(f""" 
         Calculator BY Batch 14 SMIT 
        _____________________________
        
        1st NUMBER :    {num1}
        2nd NUMBER :    {num2}
        Operator   :    {op}
                _____________
            YOUR ANSWER  : {answer}
        
        
        _____________________________""")
```
