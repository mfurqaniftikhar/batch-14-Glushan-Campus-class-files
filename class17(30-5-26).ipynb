```python
class laptop:# blue print
    # initinzler/ consturucture
    def __init__(self,brand,windows,ram,memory,processor,grafic_card):
        self.brand = brand
        self.windows = windows
        self.ram = ram
        self.memory = memory
        self.processor = processor
        self.grafic_card = grafic_card
    
    def watching_movies(self):
        print(f"watching movies on {self.brand}")
    def coding(self):
        print(f"gpt sy krwa lia code toh {self.brand}" )
    def playing_games(self):
        print(f"playing gamess... {self.brand}")
    def editing(self):
        print(f"editing videos...{self.brand}")
    def desgingin(self):
        print(f"desging.....{self.brand}")
```
